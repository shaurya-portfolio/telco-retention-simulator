### Handling TotalCharges

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'],errors='coerce')
print(df['TotalCharges'].isnull().sum())
```

```python
df.dropna(inplace=True)
```

```python
df.shape
```

### Dropping and Encoding


```python
df.drop('customerID',axis=1,inplace=True)
```

```python
# encoding the target variable
df['Churn'] = df['Churn'].map({'Yes':1,"No":0})
```

```python
# one hot encoding for all other categorical variable
df = pd.get_dummies(df,drop_first=True,dtype=int)
```

```python
df
```

### Clean Column Names

```python
df.columns = df.columns.str.replace(' ', '_').str.replace('(', '').str.replace(')', '').str.replace('-', '_')
```

## Data Partitioning

```python
df_historical = df.iloc[:5000]
df_live = df.iloc[5000:]
```

```python
df_historical.to_csv("../data/historical_train_data.csv",index=False)
df_live.to_csv('../data/future_live_data.csv', index=False)
```
